## EDA (Easy Data augmentation)

```python
# Function to perform data augmentation for EDA and save to CSV
def perform_augmentation_eda(percentage, train_df, output_path):
    # Initialize EDA augmenter
    eda_aug = EasyDataAugmenter()

    # Select a random subset of the data
    sample_df = train_df.sample(frac=percentage, random_state=42)

    # Augment the selected data
    sample_df['text'] = sample_df['text'].apply(lambda x: eda_aug.augment(x)[0])

    # Append augmented data to the original data
    augmented_train_df = pd.concat([train_df, sample_df])

    # Save the DataFrame to a CSV file
    augmented_train_df.to_csv(output_path, index=False)

    return augmented_train_df
```
